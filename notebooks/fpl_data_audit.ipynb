```python
import pandas as pd

#read the data file

df = pd.read_csv("../data/players_2025_26.csv")

df.shape
```

```python
# count how many values are missing in each column
missing = df.isna().sum()
missing
```

```python
# sort the most missing values first
missing_sorted = missing.sort_values(ascending=False)
missing_sorted
```

```python
# show only the columns that have missing values
missing[missing > 0]
```

```python
#How many duplicated rows
df.duplicated().sum()
```

```python
#How many rows have players with 0 minutes played
(df["minutes"] == 0).sum()
```

```python
# import the plotting library
import matplotlib.pyplot as plt

# draw a histogram of the target column
df["total_points"].hist(bins=20)

# label the x-axis
plt.xlabel("total_points")

# label the y-axis
plt.ylabel("number of players")

# give the plot a title
plt.title("Distribution of total_points")

# display plot
plt.show()
```
